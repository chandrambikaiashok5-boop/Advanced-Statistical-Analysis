```python
# Select numerical columns
scores = ['math score', 'reading score', 'writing score']

# Perform normality tests
for column in scores:
    data = df[column]

    print("\nColumn:", column)

    # Shapiro-Wilk Test
    shapiro_stat, shapiro_p = stats.shapiro(data)

    print("Shapiro-Wilk Test")
    print("Statistic:", shapiro_stat)
    print("P-value:", shapiro_p)

    # Kolmogorov-Smirnov Test
    standardized = (data - data.mean()) / data.std(ddof=1)
    ks_stat, ks_p = stats.kstest(standardized, 'norm')

    print("Kolmogorov-Smirnov Test")
    print("Statistic:", ks_stat)
    print("P-value:", ks_p)
```

```python
# Separate math scores by gender
male = df[df['gender'] == 'male']['math score']
female = df[df['gender'] == 'female']['math score']

# Independent two-sample t-test (Welch's test)
t_stat, p_value = stats.ttest_ind(male, female, equal_var=False)

print("Two-Sample T-Test")
print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: Significant difference between groups")
else:
    print("Fail to reject H0: No significant difference")
```

```python
u_stat, p_value = stats.mannwhitneyu(
    male, female, alternative='two-sided'
)

print("Mann-Whitney U Test")
print("U-statistic:", u_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: Significant difference between groups")
else:
    print("Fail to reject H0: No significant difference")
```

```python
# Group math scores by parental education level
groups = [
    group['math score'].values
    for name, group in df.groupby('parental level of education')
]

# Perform One-Way ANOVA
f_stat, p_value = stats.f_oneway(*groups)

print("One-Way ANOVA Results")
print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: Significant difference between groups")
else:
    print("Fail to reject H0: No significant difference")
```

```python
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm

# Create Two-Way ANOVA model
model = ols(
    'Q("math score") ~ C(gender) * C(Q("parental level of education"))',
    data=df
).fit()

# Perform Two-Way ANOVA
anova_results = anova_lm(model, typ=2)

print("Two-Way ANOVA Results:")
print(anova_results)
```
